```python
df["TotalCharges"] = pd.to_numeric(
    df["TotalCharges"],
    errors="coerce"
)
```

```python
df[df["TotalCharges"].isnull()]
```

```python
df.loc[df["tenure"] == 0, "TotalCharges"] = (
    df.loc[df["tenure"] == 0, "TotalCharges"].fillna(0)
)
```

```python
print("Missing values:", df.isnull().sum().sum())
print("Duplicate records:", df.duplicated().sum())
print("Dataset shape:", df.shape)
```

```python
plt.figure(figsize=(6, 4))
sns.countplot(data=df, x="Churn", hue="Churn")
plt.title("Customer Churn Distribution")
plt.xlabel("Churn Status")
plt.ylabel("Number of Customers")
plt.show()
```

```python
churn_rate = (df["Churn"] == "Yes").mean() * 100

print(f"Churn Rate: {churn_rate:.2f}%")
```

```python
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x="Contract", hue="Churn")
plt.title("Churn by Contract Type")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x="tenure", hue="Churn", bins=30)
plt.title("Customer Churn by Tenure")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x="Churn", y="MonthlyCharges")
plt.title("Monthly Charges vs Churn")
plt.show()
```

```python
plt.figure(figsize=(10, 5))
sns.countplot(data=df, x="PaymentMethod", hue="Churn")
plt.title("Churn by Payment Method")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
df["TenureGroup"] = pd.cut(
    df["tenure"],
    bins=[-1, 12, 24, 48, 72],
    labels=["New", "Regular", "Loyal", "Long-Term"]
)

df[["tenure", "TenureGroup"]].head()
```

```python
df["AverageMonthlySpend"] = (
    df["TotalCharges"] / df["tenure"].clip(lower=1)
)

df[["TotalCharges", "tenure", "AverageMonthlySpend"]].head()
```

```python
services = [
    "PhoneService",
    "MultipleLines",
    "OnlineSecurity",
    "OnlineBackup",
    "DeviceProtection",
    "TechSupport",
    "StreamingTV",
    "StreamingMovies"
]

df["TotalServices"] = (df[services] == "Yes").sum(axis=1)

df[["customerID", "TotalServices"]].head()
```

```python
df["Churn"] = df["Churn"].map({
    "Yes": 1,
    "No": 0
})

df["Churn"].value_counts()
```

```python
df.to_csv("../data/processed/cleaned_telco.csv", index=False)

print("Processed dataset saved successfully!")
print("Dataset shape:", df.shape)
```
